```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import Markdown, display

repo_root = Path.cwd().resolve()
while repo_root != repo_root.parent and not (repo_root / 'src').exists():
    repo_root = repo_root.parent

if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
if str(repo_root / 'src') not in sys.path:
    sys.path.insert(0, str(repo_root / 'src'))

from crash_tests.spikeforest_audit import SpikeForestAuditConfig, run_with_config
```

```python
RUN_NAME = 'spikeforest_audit_v1'

config = SpikeForestAuditConfig(
    parquet_path=repo_root / 'data/raw/33000_ROWS.parquet',
    extraction_notebook=repo_root / 'notebooks/Units_Spikeforest_Extraction_Colab.ipynb',
    output_root=repo_root / 'crash_tests/spikeforest_audit/outputs',
    run_name=RUN_NAME,
    verbose=True,
)

config
```

```python
study_manifest = pd.read_csv(output_dir / 'study_manifest.csv')
recording_manifest = pd.read_csv(output_dir / 'recording_manifest.csv')
missing_items = pd.read_csv(output_dir / 'missing_items.csv')

display(Markdown('## Study Manifest'))
display(study_manifest)

display(Markdown('## Recording Coverage Gaps'))
display(recording_manifest.sort_values(['missing_common_sorter_count', 'study_set', 'recording_name'], ascending=[False, True, True]).head(25))

display(Markdown('## Missing Items'))
display(missing_items)
```
