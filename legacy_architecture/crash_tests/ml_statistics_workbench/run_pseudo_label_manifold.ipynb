# Real-x / Pseudo-y Manifold Benchmark

This notebook uses real unlabeled paired units as target-domain `x`, discovers manifold structure on the non-test paired pool, pseudo-labels only high-confidence in-support rows, and retrains context-first tree models with weighted pseudo labels.

```python
from pathlib import Path
import sys

from IPython.display import display

repo_root = Path.cwd().parents[1] if Path.cwd().name == 'ml_statistics_workbench' else Path('.').resolve()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
if str(repo_root / 'src') not in sys.path:
    sys.path.insert(0, str(repo_root / 'src'))

from crash_tests.ml_statistics_workbench.pseudo_label_manifold import (
    PseudoLabelManifoldConfig,
    run_with_config,
)
```

```python
RUN_NAME = 'pseudo_label_manifold_full_balanced_v2'
SMOKE_RUN = False

config = PseudoLabelManifoldConfig(
    parquet_path=repo_root / 'data/raw/33000_ROWS.parquet',
    output_root=repo_root / 'crash_tests/ml_statistics_workbench/outputs',
    run_name=RUN_NAME,
    max_pseudo_multiplier=1.5,
    max_pseudo_per_labeled_ratio=1.5,
    pseudo_weight_scale=0.25,
    pseudo_weight_floor=0.08,
    verbose=True,
)

if SMOKE_RUN:
    config.max_pseudo_multiplier = 0.9
    config.max_pseudo_per_labeled_ratio = 1.0
    config.max_pseudo_per_recording = 20
    config.max_pseudo_per_cluster = 48

config
```
