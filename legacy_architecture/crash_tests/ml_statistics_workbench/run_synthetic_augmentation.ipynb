# Synthetic Paired Augmentation

This notebook runs a leakage-safe synthetic paired augmentation benchmark on the fixed paired holdout.

```python
from pathlib import Path
import sys

from IPython.display import display

repo_root = Path.cwd().parents[1] if Path.cwd().name == 'ml_statistics_workbench' else Path('.').resolve()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
if str(repo_root / 'src') not in sys.path:
    sys.path.insert(0, str(repo_root / 'src'))

from crash_tests.ml_statistics_workbench.synthetic_augmentation import (
    SyntheticAugmentationConfig,
    run_with_config,
)
```

```python
RUN_NAME = 'synthetic_augmentation_full_v1'
SMOKE_RUN = False

config = SyntheticAugmentationConfig(
    parquet_path=repo_root / 'data/raw/33000_ROWS.parquet',
    output_root=repo_root / 'crash_tests/ml_statistics_workbench/outputs',
    run_name=RUN_NAME,
    budgets=(40, 60, 80, 107),
    repeats=3,
    augmentation_factor=1.0,
    augmentation_weight=0.35,
    augmentation_noise_scale=0.01,
    augmentation_k_neighbors=3,
    subgroup_column='study_set',
    verbose=True,
)

if SMOKE_RUN:
    config.budgets = (40, 107)
    config.repeats = 2
    config.augmentation_factor = 0.75

config
```

```python
output_dir = run_with_config(config)
output_dir
```
