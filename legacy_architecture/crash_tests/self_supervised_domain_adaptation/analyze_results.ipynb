# SSL + Context Domain Adaptation Analysis

Inspect the latest crash-test run artifacts.

```python
from pathlib import Path
import json

import pandas as pd
from IPython.display import display

cwd = Path.cwd().resolve()
repo_root = cwd.parents[1] if cwd.name == 'self_supervised_domain_adaptation' else cwd
output_root = repo_root / 'crash_tests' / 'self_supervised_domain_adaptation' / 'outputs'
latest = sorted([p for p in output_root.iterdir() if p.is_dir()])[-1]
print('Latest run:', latest)
latest
```

```python
config = json.loads((latest / 'config.json').read_text())
split_manifest = json.loads((latest / 'split_manifest.json').read_text())
metrics = pd.read_csv(latest / 'metrics.csv')
predictions = pd.read_csv(latest / 'predictions.csv')
baselines = pd.read_csv(latest / 'baseline_comparison.csv')
stress = pd.read_csv(latest / 'stress_test_results.csv')

display(pd.DataFrame([config]))
display(pd.DataFrame([split_manifest['main']]))
```

```python
print('Main benchmark results')
display(metrics.sort_values(['target', 'mae', 'model_id']))

print('Baseline comparison')
display(baselines.sort_values(['target', 'mae', 'model_id']))

print('Contextual vs unit-only SSL models')
ssl_only = metrics[metrics['family'] == 'ssl'].copy()
display(ssl_only.sort_values(['target', 'mae', 'model_id']))
```

```python
if len(stress):
    print('Stress test results')
    display(stress.sort_values(['held_out_study', 'target', 'mae', 'model_id']))
else:
    print('No stress-test results found in this run.')

print('Prediction sample')
display(predictions.head(20))
```
