# MMD Domain Adaptation Analysis

```python
from pathlib import Path
import json

import pandas as pd
from IPython.display import display

cwd = Path.cwd().resolve()
repo_root = cwd.parents[1] if cwd.name == 'mmd_domain_adaptation' else cwd
output_root = repo_root / 'crash_tests' / 'mmd_domain_adaptation' / 'outputs'
latest = sorted([p for p in output_root.iterdir() if p.is_dir()])[-1]
print('Latest run:', latest)
latest
```

```python
config = json.loads((latest / 'config.json').read_text())
split_manifest = json.loads((latest / 'split_manifest.json').read_text())
metrics = pd.read_csv(latest / 'metrics.csv')
predictions = pd.read_csv(latest / 'predictions.csv')
baselines = pd.read_csv(latest / 'baseline_comparison.csv')
lambda_sweep = pd.read_csv(latest / 'lambda_sweep.csv')
history = pd.read_csv(latest / 'training_history.csv')

display(pd.DataFrame([config]))
display(pd.DataFrame([split_manifest['main']]))
```

```python
print('Holdout results')
display(metrics.sort_values(['target', 'mae', 'variant_id']))

print('Lambda sweep')
display(lambda_sweep.sort_values(['target', 'eval_stage', 'cv_mae_mean', 'lambda']))
```

```python
print('Baseline comparison')
display(baselines.sort_values(['target', 'mae', 'variant_id']))

print('Prediction sample')
display(predictions.head(20))

print('Training history tail')
display(history.tail(60))
```
